# Loaded all 4 station CSV Files into SQLite database

```python
import pandas as pd
import sqlite3
import os

# Folder paths and database path

DATA_RAW = r"D:\Varanasi_Air_Quality\Data\Raw"
DB_PATH  = r"D:\Varanasi_Air_Quality\sql\ air_quality.db"

# Creating Dictionary

stations = {
    "bhelupur"     : "bhelupur.csv",
    "ardhali_bazar": "ardhali_bazar.csv",
    "bhu_iesd"     : "bhu_iesd.csv",
    "maldahiya"    : "maldahiya.csv",
}

# It connects to SQLite Database

conn = sqlite3.connect(DB_PATH)
print("Database connected")

# Build the full path to each CSV file, read it into a DataFrame, 
# convert the 'Timestamp' column to datetime format, and then load it into the SQLite database. 

for station_name, filename in stations.items():
    filepath = os.path.join(DATA_RAW, filename)
    df = pd.read_csv(filepath)
    df['Timestamp'] = pd.to_datetime(df['Timestamp'])
    df.to_sql(station_name, conn, if_exists='replace', index=False)
    print(f"{station_name} loaded → {len(df)} rows")

conn.close()
print("\nAll 4 stations loaded into air_quality.db ")
```

#  Creating a dictionary to store the cleaned DataFrames for each station

```python


conn = sqlite3.connect(DB_PATH)


dfs = {}
for name, file in stations.items():
    dfs[name] = pd.read_csv(os.path.join(DATA_RAW ,f"{file}"))

# Basic info of all 4 stations

for name in dfs.keys():
    print(f"\n{'='*50}")
    print(f"  STATION : {name.upper()}")
    print(f"{'='*50}")
    print("Rows     : ", dfs[name].shape[0])
    print("Columns  : ", dfs[name].shape[1])
    print("\nFeatures : \n", dfs[name].columns.tolist())
    print("\nMissing values : \n", dfs[name].isnull().sum())
    print("\nUnique values  : \n", dfs[name].nunique())
    print("\nData types     : \n", dfs[name].dtypes)

conn.close()
    
```

# drop useless columns

```python
conn = sqlite3.connect(DB_PATH)

cols_to_drop = [
    'Toluene (µg/m³)', 'Xylene (µg/m³)', 'O Xylene (µg/m³)',
    'BP (mmHg)', 'VWS (m/s)', 'NOx (ppb)', 'NO2 (µg/m³)',
    'NH3 (µg/m³)', 'Eth-Benzene (µg/m³)', 'MP-Xylene (µg/m³)',
    'RF (mm)', 'WD (deg)'
]

for name, file in stations.items():
    dfs[name].drop(columns=cols_to_drop, inplace=True, errors='ignore')
    print(f"-> {name}: {dfs[name].shape[1]} columns remaining")

conn.close()
```

# Add Time Feature column

```python


#  function for time of day 

def time_of_day(h):
    if   5 <= h < 12 : return 'Morning'
    elif 12 <= h < 17 : return 'Afternoon'
    elif 17 <= h < 21 : return 'Evening'
    else              : return 'Night'

# Now apply to all 4 stations

for name in dfs.keys():
    dfs[name]['Timestamp'] = pd.to_datetime(dfs[name]['Timestamp'])
    dfs[name]['hour']        = dfs[name]['Timestamp'].dt.hour
    dfs[name]['day']         = dfs[name]['Timestamp'].dt.day
    dfs[name]['month']       = dfs[name]['Timestamp'].dt.month
    dfs[name]['month_name']  = dfs[name]['Timestamp'].dt.strftime('%b')
    dfs[name]['day_of_week'] = dfs[name]['Timestamp'].dt.day_name()
    dfs[name]['is_weekend']  = dfs[name]['Timestamp'].dt.dayofweek >= 5

    dfs[name]['season'] = dfs[name]['month'].map({
        12: 'Winter',       1: 'Winter',        2: 'Winter',
         3: 'Summer',       4: 'Summer',         5: 'Summer',
         6: 'Monsoon',      7: 'Monsoon',        8: 'Monsoon',
         9: 'Monsoon',
        10: 'Post-Monsoon', 11: 'Post-Monsoon'
    })

    dfs[name]['time_of_day'] = dfs[name]['hour'].apply(time_of_day)

    print(f" {name} — time features added")

print("\nPreview :")
print(dfs['bhelupur'][['Timestamp','hour','season','day_of_week','time_of_day','is_weekend']].head(10))
```

# Handle Missing Values in PM2.5

```python

# Strategy : Linear interpolation for gaps of 3 hours or less
#            Leave larger gaps as NaN (do not guess)

for name in dfs.keys():

    # Count missing before
    before = dfs[name]['PM2.5 (µg/m³)'].isnull().sum()

    # Linear interpolation — fills gaps of maximum 3 consecutive missing hours
    dfs[name]['PM2.5 (µg/m³)'] = dfs[name]['PM2.5 (µg/m³)'].interpolate(
        method = 'linear',
        limit  = 3,                    # maximum 3 consecutive NaN to fill
        limit_direction = 'forward'    # fill in forward direction
    )

    # Count missing after
    after = dfs[name]['PM2.5 (µg/m³)'].isnull().sum()

    print(f"{name}")
    print(f"   Missing before : {before}")
    print(f"   Missing after  : {after}")
    print(f"   Filled         : {before - after} values")
    print()
```

# Handle Missing Values in Ozone

```python

for name in dfs.keys():

    before = dfs[name]['Ozone (µg/m³)'].isnull().sum()

    dfs[name]['Ozone (µg/m³)'] = dfs[name]['Ozone (µg/m³)'].interpolate(
        method          = 'linear',
        limit           = 3,
        limit_direction = 'forward'
    )

    after = dfs[name]['Ozone (µg/m³)'].isnull().sum()

    print(f"  {name}")
    print(f"   Missing before : {before}")
    print(f"   Missing after  : {after}")
    print(f"   Filled         : {before - after} values")
    print()
```

#  Handle Missing Values in PM10 (2.3% missing)

```python

for name in dfs.keys():
    before = dfs[name]['PM10 (µg/m³)'].isnull().sum()
    dfs[name]['PM10 (µg/m³)'] = dfs[name]['PM10 (µg/m³)'].interpolate(
        method='linear', limit=3, limit_direction='forward'
    )
    after = dfs[name]['PM10 (µg/m³)'].isnull().sum()
    print(f"{name} | Before : {before} | After : {after} | Filled : {before - after}")
```

# Handle Missing Values in  CO (high fluctuation)

```python
for name in dfs.keys():
    before = dfs[name]['CO (mg/m³)'].isnull().sum()
    dfs[name]['CO (mg/m³)'] = dfs[name]['CO (mg/m³)'].interpolate(
        method='linear', limit=3, limit_direction='forward'
    )
    after = dfs[name]['CO (mg/m³)'].isnull().sum()
    print(f" {name} | Before : {before} | After : {after} | Filled : {before-after}")
```

# Handle Missing Values in SO2
Fill with monthly median — more accurate than overall median
because SO2 varies slightly by season

```python
for name in dfs.keys():
    before = dfs[name]['SO2 (µg/m³)'].isnull().sum()
    dfs[name]['SO2 (µg/m³)'] = dfs[name].groupby('month')['SO2 (µg/m³)'].transform(
        lambda x: x.fillna(x.median())
    )

    after = dfs[name]['SO2 (µg/m³)'].isnull().sum()
    print(f" {name} | Before : {before} | After : {after} | Filled : {before-after}")
```

# Handle Missing Valus in NO 
interpolate first, then fill remaining with hourly median

```python
for name in dfs.keys():
    before = dfs[name]['NO (µg/m³)'].isnull().sum()

    # Step 1 — interpolate small gaps
    dfs[name]['NO (µg/m³)'] = dfs[name]['NO (µg/m³)'].interpolate(
        method='linear', limit=3, limit_direction='forward'
    )

    # Step 2 — fill remaining large gaps with hourly median
    hourly_median = dfs[name].groupby('hour')['NO (µg/m³)'].transform('median')
    dfs[name]['NO (µg/m³)'] = dfs[name]['NO (µg/m³)'].fillna(hourly_median)

    after = dfs[name]['NO (µg/m³)'].isnull().sum()
    print(f"✓ {name} | Before : {before} | After : {after} | Filled : {before-after}")
```

# Handle Missing Valus in Benzene 
interpolate first, then fill remaining with hourly median

```python
for name in dfs.keys():
    before = dfs[name]['Benzene (µg/m³)'].isnull().sum()

    dfs[name]['Benzene (µg/m³)'] = dfs[name]['Benzene (µg/m³)'].interpolate(
        method='linear', limit=3, limit_direction='forward'
    )

    hourly_median = dfs[name].groupby('hour')['Benzene (µg/m³)'].transform('median')
    dfs[name]['Benzene (µg/m³)'] = dfs[name]['Benzene (µg/m³)'].fillna(hourly_median)

    after = dfs[name]['Benzene (µg/m³)'].isnull().sum()
    print(f" {name} | Before : {before} | After : {after} | Filled : {before-after}")
```

#  Merge all 4 stations into one master DataFrame

```python

# Add station name column to each DataFrame

for name in dfs.keys():
    dfs[name]['station'] = name

# Stack all 4 DataFrames vertically into one master

master_df = pd.concat(dfs.values(), ignore_index=True)

# Sort by station then timestamp

master_df = master_df.sort_values(['station', 'Timestamp']).reset_index(drop=True)


print("Master DataFrame created ")
print(f"Total rows    : {len(master_df):,}")
print(f"Total columns : {master_df.shape[1]}")
print(f"\nRows per station :")
print(master_df.groupby('station').size())
print(f"\nColumns :")
print(master_df.columns.tolist())
```

# Save master DataFrame to SQLite and CSV

```python
# ---- Save to SQLite ----
conn = sqlite3.connect(DB_PATH)

master_df.to_sql('master', conn, if_exists='replace', index=False)

# Verify with a quick SQL query

test = pd.read_sql_query("""
    SELECT station,
           COUNT(*) as total_rows
    FROM master
    GROUP BY station
""", conn)

conn.close()

print(" Master table saved to SQLite")
print()
print(test.to_string(index=False))

# ---- Save to CSV ----
master_df.to_csv('D:/Varanasi_Air_Quality/data/processed/master_clean.csv', index=False)
print()
print("Master CSV saved → D:/Varanasi_Air_Quality/Data/Proccessed/master_clean.csv")
```
